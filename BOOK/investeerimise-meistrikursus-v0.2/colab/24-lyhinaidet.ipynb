## 1. Liitintress

Kui suureks kasvab 1000 eurot viie aastaga 6% aastase tootluse korral?

Iga järgmine tootlus teenitakse ka varasema tootluse pealt. Tegemist on konstantse tootluse mudeliga.

Piirangud: 6% sisesta 0.06, mitte 6. Sissemaksed ja maksud puuduvad.

Ülesanne: Lisa 100-eurone sissemakse iga aasta lõpus.

```python
algus, tootlus, aastad = 1000, 0.06, 5
lopp = algus * (1 + tootlus) ** aastad
print(f"Lõppväärtus: {lopp:.2f} eurot")
```

## 2. Reaaltootlus

Nominaaltootlus on 8% ning hinnatase tõuseb 5%. Kui palju kasvab ostujõud?

Nominaalset kasvutegurit tuleb jagada hinnataseme kasvuteguriga.

Piirangud: 8−5=3% on ligikaudne, mitte täpne vastus. Mõlemad määrad peavad katma sama perioodi.

Ülesanne: Arvuta reaaltootlus 4% nominaaltootluse ja 6% inflatsiooni korral.

```python
nominaal, inflatsioon = 0.08, 0.05
reaal = (1 + nominaal) / (1 + inflatsioon) - 1
print(f"Reaaltootlus: {reaal:.4%}")
```

## 3. Kahjumist taastumine

Kui palju peab allesjäänud kapital kasvama, et taastada algväärtus?

Protsendibaas pärast kahjumit on väiksem.

Piirangud: 100% kahjumi järel ei saa sama nullkapitaliga taastuda; valemis on siis jagamine nulliga.

Ülesanne: Leia taastumiseks vajalik tootlus 35% kahjumi järel.

```python
for kahjum in (0.10, 0.25, 0.50, 0.80):
    vajalik = kahjum / (1 - kahjum)
    print(f"Kahjum {kahjum:.0%}: taastumiseks {vajalik:.2%}")
```

## 4. Aritmeetiline keskmine ja CAGR

Mida tähendab +20% ja −20% kahe aasta tootlus?

Aritmeetiline keskmine on 0%, aga kapital väheneb.

Piirangud: Kahe aasta kogutootlus ja aastane CAGR pole sama.

Ülesanne: Võrdle tootlustega +10% ja −10%.

```python
from math import prod
r = [0.20, -0.20]
aritmeetiline = sum(r) / len(r)
kasvutegur = prod(1 + x for x in r)
cagr = kasvutegur ** (1 / len(r)) - 1
print(f"Aritmeetiline: {aritmeetiline:.2%}")
print(f"Kogutootlus: {kasvutegur-1:.2%}; CAGR: {cagr:.4%}")
```

Diskonteerimine küsib, milline tänane summa kasvaks eeldatud määraga tulevase rahavooni.

## 6. Võlakirja hind

Leia 1000-eurose, 5% aastakupongiga ja kolmeaastase võlakirja hind 6% tootlusnõude korral.

Hind on kõigi lepinguliste maksete nüüdisväärtuste summa.

Piirangud: Eeldame maksehäire puudumist ja ostu kupongipäeval; kogunenud intress puudub.

Ülesanne: Kontrolli, et kupongiga võrdse tootlusnõude korral on hind nimiväärtus.

```python
nimivaartus, kupong, tootlus, aastad = 1000, 0.05, 0.06, 3
hind = sum(nimivaartus * kupong / (1+tootlus)**t
           for t in range(1, aastad+1))
hind += nimivaartus / (1+tootlus)**aastad
print(f"Hind: {hind:.4f}")
```

## 7. Duration ja intressitundlikkus

Hinda sama võlakirja tundlikkust väikese tootlusnõude muutuse suhtes.

Macaulay duration kaalub maksete aegu nüüdisväärtusega; modified duration annab lokaalse hinnatundlikkuse.

Piirangud: Duration ei ole täpne suurte šokkide ega optsioonidega võlakirjade korral.

Ülesanne: Võrdle 1-protsendipunktilist šokki täpse hinnaarvutusega.

```python
F, c, y, n = 1000, 0.05, 0.06, 3
cf = [F*c] * n
cf[-1] += F
pv = [v/(1+y)**t for t,v in enumerate(cf,1)]
P = sum(pv)
d_mac = sum(t*v for t,v in enumerate(pv,1))/P
d_mod = d_mac/(1+y)
print(f"Macaulay: {d_mac:.4f}; modified: {d_mod:.4f}")
print(f"Ligikaudne hinnamuutus +0.01 šokiga: {-d_mod*.01:.4%}")
```

## 9. Korrelatsioon tootlustest

Leia kahe väikese tootlusrea Pearsoni korrelatsioon.

Korrelatsioon mõõdab standardiseeritud lineaarset koosliikumist.

Piirangud: Hinnatasemete korrelatsioon ei asenda tootluste korrelatsiooni; konstantse rea korral on näitaja määramata.

Ülesanne: Asenda teine rida esimese vastandarvudega.

```python
from statistics import mean
from math import sqrt
a, b = [.01,-.02,.03,.00], [.02,-.01,.04,.01]
da, db = [x-mean(a) for x in a], [x-mean(b) for x in b]
rho = sum(x*y for x,y in zip(da,db))/sqrt(
    sum(x*x for x in da)*sum(y*y for y in db))
print(f"Korrelatsioon: {rho:.4f}")
```

## 10. Kahe vara standardhälve

Arvuta 50/50 portfelli standardhälve, kui mõlema vara oma on 20% ning korrelatsioon 0.

Kovariatsioon määrab, kui palju üks vara teise kõikumisi tasakaalustab.

Piirangud: Ära keskmista standardhälbeid ega unusta ruutjuurt.

Ülesanne: Korda korrelatsiooniga 1 ja −1.

```python
from math import sqrt
w, s1, s2, rho = .5, .2, .2, 0
var = w*w*s1*s1 + (1-w)**2*s2*s2 + 2*w*(1-w)*rho*s1*s2
print(f"Portfelli standardhälve: {sqrt(var):.4%}")
```

## 11. Sharpe’i suhtarv

Arvuta nelja aastase ületootluse põhjal Sharpe’i valimihinnang.

Näide kasutab sama perioodi riskivaba tootluse suhtes saadud ületootlusi.

## 12. Maksimaalne langus

Leia vara suurim tipust põhjani langus, arvestades algkapitali.

Iga väärtust võrreldakse senise suurima väärtusega.

Piirangud: Kui algkapital puudub reast, võib esimene kahjum jääda arvestamata; raha sisse- ja väljamaksed vajavad kohandust.

Ülesanne: Lisa rea lõppu väärtus 70.

```python
vaartused = [100, 120, 90, 110, 130]
tipp, suurim = vaartused[0], 0
for v in vaartused:
    tipp = max(tipp,v)
    suurim = min(suurim,v/tipp-1)
print(f"Maksimaalne langus: {suurim:.2%}")
```

## 13. Monte Carlo keskmise kontroll

Võrdle normaalse logtootlusega ühe aasta lõppväärtuse simulatsiooni teoreetilise ootusega.

Normaaljaotus kirjeldab siin logtootlust, nii et väärtus jääb positiivseks.

Piirangud: Simulatsioon pole turuennustus; valitud mu ja sigma on eeldused.

Ülesanne: Suurenda korduste arvu ja võrdle keskmise viga.

```python
from random import Random
from math import exp
from statistics import mean
rng = Random(42)
mu, sigma, algus = .06, .20, 1000
vaartused = [algus*exp(rng.gauss(mu,sigma)) for _ in range(20000)]
print(f"Simulatsiooni keskmine: {mean(vaartused):.2f}")
print(f"Teoreetiline ootus: {algus*exp(mu+sigma*sigma/2):.2f}")
```

## 19. Beeta ja alfa

Leia OLS-seos kahe lühikese ületootlusrea vahel.

Beeta mõõdab tundlikkust valitud turuvõrdluse suhtes; alfa on regressiooni vabaliige.

Piirangud: Nelja vaatlusega täiuslik seos on õpetlik konstruktsioon, mitte tõend juhioskuse kohta.

Ülesanne: Lisa ühele fonditootlusele müra.

```python
from statistics import mean
x=[-.02,.00,.02,.04]
y=[-.02,.01,.04,.07]
beta=sum((a-mean(x))*(b-mean(y)) for a,b in zip(x,y))/sum((a-mean(x))**2 for a in x)
alfa=mean(y)-beta*mean(x)
print(f"Beeta: {beta:.4f}; perioodi alfa: {alfa:.4%}")
```

## 21. Bayesi uuendus

Hea mudeli eelnev tõenäosus on 10%; testi läbimise tõenäosus hea mudeli korral 80%, halva korral 20%.

Positiivse testi tõendusjõud sõltub ka headuse eelnevast tõenäosusest.

Piirangud: Testi täpsus ja mudeli headuse tõenäosus pärast testi pole sama; sisendmäärad on siin eeldused.

Ülesanne: Muuda hea mudeli eelnev tõenäosus 1%-ks.

```python
p_hea,p_pos_hea,p_pos_halb=.10,.80,.20
p_pos=p_hea*p_pos_hea+(1-p_hea)*p_pos_halb
posterior=p_hea*p_pos_hea/p_pos
print(f"P(hea | positiivne test): {posterior:.4%}")
```

## 22. Kelly panuse osakaal

Võida panustatud euro kohta euro tõenäosusega 55%, muidu kaota panus.

Kelly maksimeerib selle binaarse kordusmudeli oodatavat logkasvu.

Piirangud: Tõenäosuse hindamisviga, sõltuvad katsed ja suured panused võivad tulemuse rikkuda; see pole isiklik panusesoovitus.

Ülesanne: Muuda tegelik võidutõenäosus 48%-ks, hoides esialgset panust.

```python
from math import log
p,b=.55,1
f=(b*p-(1-p))/b
g=p*log(1+b*f)+(1-p)*log(1-f)
print(f"Kelly osakaal: {f:.2%}; oodatav logkasv: {g:.6f}")
```

## 23. Lihtne viitajaga järeltest

Kas eelmise perioodi positiivsele tootlusele järgnev hoidmine võidab alati osta-ja-hoia strateegiat?

Positsioon perioodil t põhineb ainult perioodil t−1 teada olnud tootlusel.

Piirangud: Lühike konstrueeritud rida ei tõenda strateegia headust. Kulu võetakse iga ühikulise positsioonimuutuse eest, lõpp-likvideerimise kulu pole lisatud.

Ülesanne: Lisa andmelekkega variant ja selgita, miks selle tulemus pole lubatav.

```python
from math import prod
r=[.02,-.01,.03,-.02,.01]
pos=[0]+[int(x>0) for x in r[:-1]]
eelmine,kasv=0,1
for w,x in zip(pos,r):
    kulu=.001*abs(w-eelmine)
    kasv*=1+w*x-kulu
    eelmine=w
print(f"Strateegia: {kasv-1:.4%}; osta-ja-hoia: {prod(1+x for x in r)-1:.4%}")
```

## 24. Regressioon versus naiivne prognoos

Kas sirgmudel annab ajaliselt eraldatud kontrollandmetel väiksema MSE kui nullmuutuse prognoos?

Treening kasutab ainult esimest nelja vaatlust; test on eraldi. Sihtmuutuja on järgmise perioodi tootlus.

Piirangud: See pole aktsiaturu prognoositavuse tõend. Ka parem MSE ei taga kuludejärgset kauplemistulu.

Ülesanne: Muuda kontrollperioodil seose märk ja võrdle tulemusi.

```python
from statistics import mean
x_train=[-2,-1,1,2]
y_train=[-.02,-.01,.01,.02]
b=sum(x*y for x,y in zip(x_train,y_train))/sum(x*x for x in x_train)
x_test,y_test=[-3,3],[-.025,.025]
mse_model=mean((b*x-y)**2 for x,y in zip(x_test,y_test))
mse_naive=mean(y*y for y in y_test)
print(f"Mudeli MSE: {mse_model:.6f}; nullprognoosi MSE: {mse_naive:.6f}")
```
