# Eesti OSINT-analüütiku viis päeva
Ainult sünteetiline õppeprojekt. Laadi üles kaasasolev ZIP. Põhirakendus ei vaja väliseid teenuseid.

```python
from google.colab import files
uploaded = files.upload()
```

```python
from pathlib import Path
import zipfile, os
base = Path("/content/osint")
base.mkdir(parents=True, exist_ok=True)
zip_path = next(Path(n) for n in uploaded if n.endswith(".zip"))
with zipfile.ZipFile(zip_path) as z:
    for name in z.namelist():
        if not (base/name).resolve().is_relative_to(base.resolve()):
            raise ValueError("Unsafe ZIP path")
    z.extractall(base)
root = next(base.rglob("src/assistant.py")).parents[1]
os.chdir(root)
print(root)
```
