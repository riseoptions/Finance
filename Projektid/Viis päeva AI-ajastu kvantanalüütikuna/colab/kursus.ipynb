# Viis päeva AI-ajastu kvantanalüütikuna

Laadi üles kursuse ZIP. Kasutatakse sünteetilist andmestikku. See märkmik käivitab paketi sama koodi; Colabi üleslaadimisdialoogi pole koostamiskeskkonnas läbi proovitud.

```python
from google.colab import files
import io, zipfile
from pathlib import Path
uploaded = files.upload()
name = next(n for n in uploaded if n.endswith('.zip'))
base = Path('/content/kvandikursus')
base.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(uploaded[name])) as z:
    for member in z.infolist():
        target = (base / member.filename).resolve()
        if not target.is_relative_to(base.resolve()):
            raise ValueError('ZIP sisaldab lubamatut rada')
    z.extractall(base)
root = base / 'viis-paeva-ai-kvandina'
assert (root/'python/projekt.py').exists()
```

```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(root/'requirements.txt')], check=True)
subprocess.run([sys.executable, str(root/'python/projekt.py')], check=True)
```
